# Working com Charts, Graphs, e Slide Decks
Claude is highly capable of working com charts, graphs, e broader slide decks. Depending on your use case, ali are a número of tips e tricks aquele you may want para take advantage of. este recipe will show you common patterns for using Claude com estes materials.

## Charts e Graphs
For the maioria part, using claude com charts e graphs is simples. Let's walk through como para ingest them e pass them para Claude, as well as alguns common tips para melhorar your results.

### Ingestion e calling the Claude API
The best way para pass Claude charts e graphs is para take advantage of its vision capabilities e the PDF support recurso. aquele is, give Claude a PDF document of the gráfico ou gráfico, along com a texto question sobre it.

At the moment, only `claude-3-5-sonnet-20241022` supports the PDF recurso. Since the recurso is still in beta, you will need para provide it com the `pdfs-2024-09-25` beta header.

```python
# Install and create the Anthropic client.
%pip install anthropic
```

```python
import base64
from anthropic import Anthropic
# While PDF support is in beta, you must pass in the correct beta header
client = Anthropic(default_headers={
    "anthropic-beta": "pdfs-2024-09-25"
  }
)
# For now, only claude-3-5-sonnet-20241022 supports PDFs
MODEL_NAME = "claude-3-5-sonnet-20241022"
```

```python
# Make a useful helper function.
def get_completion(messages):
    response = client.messages.create(
        model=MODEL_NAME,
        max_tokens=8192,
        temperature=0,
        messages=messages
    )
    return response.content[0].text
```

```python
# To start, we'll need a PDF. We will be using the .pdf document located at cvna_2021_annual_report.pdf.
# Start by reading in the PDF and encoding it as base64.
with open("./documents/cvna_2021_annual_report.pdf", "rb") as pdf_file:
    binary_data = pdf_file.read()
    base_64_encoded_data = base64.b64encode(binary_data)
    base64_string = base_64_encoded_data.decode('utf-8')
```

Let's see como we can pass Este documento para the model alongside a simples question.

```python
messages = [
    {
        "role": 'user',
        "content": [
            {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": base64_string}},
            {"type": "text", "text": "What's in this document? Answer in a single sentence."}
        ]
    }
]

print(get_completion(messages))
```

aquele's pretty good! agora let's ask it alguns mais useful questions.

```python
questions = [
    "What was CVNA revenue in 2020?",
    "How many additional markets has Carvana added since 2014?",
    "What was 2016 revenue per retail unit sold?"
]

for index, question in enumerate(questions):
    messages = [
        {

            "role": 'user',
            "content": [
                {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": base64_string}},
                {"type": "text", "text": question}
            ]
        }
    ]

    print(f"\n----------Question {index+1}----------")
    print(get_completion(messages))
```

As you can see, Claude is capable of answering fairly detailed questions sobre charts e graphs. However, ali are alguns tips e tricks aquele will help you get the maioria out of it.
- às vezes Claude's arithmetic capabilities get in the way. You'll notice aquele se you sample the third question acima it will occasionally saída an incorrect final answer because it messes up the arithmetic. Consider providing Claude com a calculator tool para ensure it doesn't make estes types of mistakes.
- com super complicated charts e graphs, we can ask Claude para "primeiro describe every data point you see in the document" as a way para elicit similar improvements para o que we seen in traditional Chain of Thought.
- Claude occasionally struggles com charts aquele depend on lots of colors para convey information, such as grouped bar charts com muitos groups. Asking Claude para primeiro identify the colors in your gráfico using HEX codes can boost its accuracy.

## Slide Decks
agora aquele we know Claude is a charts e graphs wizard, it is only logical aquele we extend it para the true home of charts e graphs - slide decks!

Slides represent a critical source of information for muitos domains, including financial services. While you *can* use packages like PyPDF para extract texto de slide decks, their gráfico/gráfico heavy nature frequentemente makes este a poor choice as models will struggle para access the information they actually need.

The PDF support recurso can be a great replacement as a resultado. It uses both extracted texto e vision in order quando processing PDF documents. In Esta seção we will go over como para use PDF documents in Claude para review slide decks, e como para deal com alguns common pitfalls of este approach.

The best way para get a typical slide deck into claude is para baixar it as a PDF e provide it directly para Claude.

```python
# Open the multi-page PDF document the same way we did earlier.
with open("./documents/twilio_q4_2023.pdf", "rb") as pdf_file:
    binary_data = pdf_file.read()
    base_64_encoded_data = base64.b64encode(binary_data)
    base64_string = base_64_encoded_data.decode('utf-8')
```

```python
# Now let's pass the document directly to Claude. Note that Claude will process both the text and visual elements of the document.
question = "What was Twilio y/y revenue growth for fiscal year 2023?"
content = [
    {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": base64_string}},
    {"type": "text", "text": question}
]

messages = [
    {
        "role": 'user',
        "content": content
    }
]

print(get_completion(messages))
```

este approach is a great way Para começar, e for alguns use cases offers the best desempenho. However, ali are alguns limitations.
- You can only include a total of 100 pages across todos provided documents in a requisição (we intend para increase este limit over tempo).
- se you are using slide content as part of RAG, introducing multimodal PDFs into your embeddings can cause problems

Luckily, we can take advantage of Claude's vision capabilities para get a much higher quality representation of the slide deck **in texto formulário** than normal pdf texto extraction allows.

We encontrar the best way para do este is para ask Claude para sequentially narrate the deck de iniciar para finish, passing it the atual slide e its prior narration. Let's see como.

```python
# Define a prompt for narrating our slide deck. We would adjut this prompt based on the nature of the deck, but keep the structure largely the same.
prompt = """
You are the Twilio CFO, narrating your Q4 2023 earnings presentation.

The entire earnings presentation document is provided to you.
Please narrate this presentation from Twilio's Q4 2023 Earnings as if you were the presenter. Do not talk about any things, especially acronyms, if you are not exactly sure you know what they mean.

Do not leave any details un-narrated as some of your viewers are vision-impaired, so if you don't narrate every number they won't know the number.

Structure your response like this:
<narration>
    <page_narration id=1>
    [Your narration for page 1]
    </page_narration>

    <page_narration id=2>
    [Your narration for page 2]
    </page_narration>

    ... and so on for each page
</narration>

Use excruciating detail for each page, ensuring you describe every visual element and number present. Show the full response in a single message.
"""
messages = [
    {
        "role": 'user',
        "content": [
        {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": base64_string}},
        {"type": "text", "text": prompt}
    ]
    }
]

# Now we use our prompt to narrate the entire deck. Note that this may take a few minutes to run (often up to 10).
completion = get_completion(messages)
```

```python
import re

# Next we'll parse the response from Claude using regex
pattern = r"<narration>(.*?)</narration>"
match = re.search(pattern, completion.strip(), re.DOTALL)
if match:
    narration = match.group(1)
else:
    raise ValueError("No narration available. Likely due to the model response being truncated.")
```

agora aquele we have a texto-based narration (it's far de perfect but it's pretty good), we have the ability para use este deck com any texto-only fluxo de trabalho. Including vector buscar!

As a final sanity verificar, let's ask a poucos questions of our narration-only Configuração!

```python
questions = [
    "What percentage of q4 total revenue was the Segment business line?",
    "Has the rate of growth of quarterly revenue been increasing or decreasing? Give just an answer.",
    "What was acquisition revenue for the year ended december 31, 2023 (including negative revenues)?"
]

for index, question in enumerate(questions):
    prompt = f"""You are an expert financial analyst analyzing a transcript of Twilio's earnings call.
Here is the transcript:
<transcript>
{narration}
</transcript>

Please answer the following question:
<question>
{question}
</question>"""
    messages = [
        {

            "role": 'user',
            "content": [
                {"type": "text", "text": prompt}
            ]
        }
    ]

    print(f"\n----------Question {index+1}----------")
    print(get_completion(messages))
```

Looks good! com estes techniques at your side, you are ready para iniciar applying models para gráfico e gráfico heavy content like slide decks.
