# Summarizing web página Content com Claude 3 Haiku
In este recipe, we'll learn como para fetch the content of a web página given its URL e então use Anthropic's Claude API para generate a summary of the página's content.

Let's iniciar by installing the Anthropic biblioteca.

primeiro, let's instalar the necessary libraries e Configuração our Anthropic cliente com our API key.

```python
# Import the required libraries
from anthropic import Anthropic

# Set up the Anthropic API client
client = Anthropic()
MODEL_NAME = "claude-3-haiku-20240229"
```

## Step 1: Fetching the web página Content
primeiro, we need para fetch the content of the web página using the provided URL. We'll use the requests biblioteca for este purpose.

```python
import requests

url = "https://en.wikipedia.org/wiki/96th_Academy_Awards"
response = requests.get(url)

if response.status_code == 200:
    page_content = response.text
else:
    print(f"Failed to fetch the web page. Status code: {response.status_code}")
    exit(1)
```

## Step 2: Preparing the entrada for Claude
próximo, we'll prepare the entrada for the Claude API. We'll criar a message aquele includes the página content e a prompt asking Claude para summarize it.

```python
prompt = f"<content>{page_content}</content>Please produce a concise summary of the web page content."

messages = [
    {"role": "user", "content": prompt}
]
```

## Step 3: Generating the Summary
agora, we'll call the Haiku para generate a summary of the web página content.

```python
response = client.messages.create(
    model="claude-3-haiku-20240307",
    max_tokens=1024,
    messages=messages
)

summary = response.content[0].text
print(summary)
```
